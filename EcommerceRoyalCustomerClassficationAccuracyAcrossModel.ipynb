```python
import os
import glob
import pandas as pd

from google.colab import drive
drive.mount('/content/drive')

# Define the target dataset directory
dataset_dir = '/content/drive/MyDrive/2026Projects/Brazilian ECommerce Data/dataset'

# List all CSV files in the directory
csv_files = glob.glob(os.path.join(dataset_dir, '*.csv'))
print("Found CSV files:", csv_files)

# Dynamically import all found CSV files into a dictionary of DataFrames
dataframes = {}
for filepath in csv_files:
    filename = os.path.basename(filepath)
    # Create a clean variable name from the filename
    var_name = os.path.splitext(filename)[0].replace(' ', '_').lower()
    try:
        dataframes[var_name] = pd.read_csv(filepath)
        print(f"Successfully imported '{filename}' as '{var_name}' (Shape: {dataframes[var_name].shape})")
    except Exception as e:
        print(f"Error importing {filename}: {e}")

# Display the first few rows of each imported dataset to verify contents
for name, df in dataframes.items():
    print(f"\n--- Preview of dataset: {name} ---")
    display(df.head())
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. Access experiment results from previously loaded dictionary or construct direct paths
dataset_dir = '/content/drive/MyDrive/2026Projects/Brazilian ECommerce Data/dataset'

df_res1 = pd.read_csv(os.path.join(dataset_dir, "Result 1 - Default_para_All_features.csv"))
df_res3 = pd.read_csv(os.path.join(dataset_dir, "Result 3 - hyperparameters_top_k_features.csv"))

# 2. Plot: All-Features Baseline vs Top-K Tuned Models
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Comparison: AUC Across Models
df_res1_sorted = df_res1.sort_values(by="auc", ascending=False)
sns.barplot(data=df_res1_sorted, x="auc", y="model", ax=axes[0], hue="model", palette="Blues_r", legend=False)
axes[0].set_title("Baseline ROC-AUC (All Features)", fontsize=12, fontweight="bold")
axes[0].set_xlabel("ROC-AUC")
axes[0].set_xlim(0.48, 0.65)

# Comparison: Top Tuned Models
df_res3_sorted = df_res3.sort_values(by="auc", ascending=False).head(5)
sns.barplot(data=df_res3_sorted, x="auc", y="model", ax=axes[1], hue="model", palette="crest", legend=False)
axes[1].set_title("Top Tuned Models (Feature Pruned)", fontsize=12, fontweight="bold")
axes[1].set_xlabel("ROC-AUC")
axes[1].set_xlim(0.48, 0.65)

plt.tight_layout()
plt.savefig("model_auc_benchmark.png", dpi=300)
plt.close()

# 3. Plot: Caught vs Missed Loyal Customers (Confusion Profile)
fig, ax = plt.subplots(figsize=(8, 4))
top_models = df_res3_sorted[['model', 'caught_loyal', 'missed_loyal']].set_index('model')
top_models.plot(kind='barh', stacked=True, color=['#2b83ba', '#d7191c'], ax=ax)
ax.set_title("Loyal Customer Identification (Caught vs. Missed)", fontsize=12, fontweight="bold")
ax.set_xlabel("Number of Test Customers")
plt.tight_layout()
plt.savefig("loyal_customer_tradeoff.png", dpi=300)
plt.close()

print("Artifacts generated: model_auc_benchmark.png, loyal_customer_tradeoff.png")
```
